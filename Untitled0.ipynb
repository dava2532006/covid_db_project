### 4. Implementasi Model Epidemiologi SIR (Susceptible-Infectious-Recovered)
Kami akan memodelkan kurva persebaran COVID-19 secara global menggunakan persamaan diferensial SIR dan mengestimasi parameter optimal $\beta$ (laju penularan) dan $\gamma$ (laju kesembuhan).

````python
import numpy as np
import pandas as pd
import kagglehub
from scipy.integrate import odeint
from scipy.optimize import minimize
import matplotlib.pyplot as plt

# Memastikan dataset df terdefinisi
if 'df' not in globals():
    file_path = "covid_19_clean_complete.csv"
    df = kagglehub.dataset_load(
        kagglehub.KaggleDatasetAdapter.PANDAS,
        "imdevskp/corona-virus-report",
        file_path
    )
    df['Date'] = pd.to_datetime(df['Date'])

# 1. Menyiapkan data historis aktual (Global harian)
actual_data = df.groupby('Date')[['Confirmed', 'Deaths', 'Recovered', 'Active']].sum().reset_index()
actual_data['R_actual'] = actual_data['Recovered'] + actual_data['Deaths']
actual_data['I_actual'] = actual_data['Active']

# Menentukan total populasi hipotesis N
N = actual_data['Confirmed'].max() * 1.5
y_actual = actual_data['I_actual'].values
t_days = np.arange(len(y_actual))

# 2. Definisikan Persamaan Diferensial SIR
def SIR_model(y, t, N, beta, gamma):
    S, I, R = y
    dSdt = -beta * S * I / N
    dIdt = beta * S * I / N - gamma * I
    dRdt = gamma * I
    return dSdt, dIdt, dRdt

# 3. Fungsi Loss untuk Fitting Parameter
def loss_function(parameters, t, y_actual, N):
    beta, gamma = parameters
    I0 = y_actual[0]
    R0 = actual_data['R_actual'].iloc[0]
    S0 = N - I0 - R0

    solution = odeint(SIR_model, (S0, I0, R0), t, args=(N, beta, gamma))
    I_pred = solution[:, 1]
    return np.mean((I_pred - y_actual) ** 2)

# 4. Melakukan Optimasi Parameter
initial_guess = [0.2, 0.1]
bounds = ((0.001, 2.0), (0.001, 1.0))

result = minimize(loss_function, initial_guess, args=(t_days, y_actual, N), bounds=bounds, method='L-BFGS-B')
optimal_beta, optimal_gamma = result.x

print(f"Optimasi Selesai!")
print(f"Estimated Beta (laju penularan): {optimal_beta:.4f}")
print(f"Estimated Gamma (laju pemulihan): {optimal_gamma:.4f}")
print(f"Basic Reproduction Number (R0): {optimal_beta / optimal_gamma:.2f}")
````

````python
# 5. Simulasi Hasil dengan Parameter Optimal
# Memastikan semua variabel yang dibutuhkan tersedia di global scope sebelum dijalankan
import numpy as np
from scipy.integrate import odeint
import matplotlib.pyplot as plt

I0_opt = y_actual[0]
R0_opt = actual_data['R_actual'].iloc[0]
S0_opt = N - I0_opt - R0_opt

sim_solution = odeint(SIR_model, (S0_opt, I0_opt, R0_opt), t_days, args=(N, optimal_beta, optimal_gamma))
S_sim, I_sim, R_sim = sim_solution.T

# 6. Visualisasi Hasil Fitting SIR
plt.figure(figsize=(14, 7))
plt.plot(actual_data['Date'], y_actual, 'ro', label='Kasus Aktif Aktual (I)', alpha=0.5, markersize=4)
plt.plot(actual_data['Date'], I_sim, 'r-', label='Prediksi Model SIR (I)', linewidth=2.5)
plt.plot(actual_data['Date'], R_sim, 'g--', label='Prediksi Sembuh/Meninggal Model SIR (R)', linewidth=2)

plt.title('Fitting Kurva Penyebaran COVID-19 Menggunakan Model SIR Global', fontsize=16)
plt.xlabel('Tanggal', fontsize=12)
plt.ylabel('Jumlah Kasus (Jiwa)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(fontsize=11)
plt.show()
````

### 5. Implementasi Model Epidemiologi SEIR (Susceptible-Exposed-Infectious-Recovered)
Model SEIR menambahkan satu kompartemen tambahan yaitu **Exposed (E)** untuk menangkap periode inkubasi penyakit sebelum penderita menjadi aktif menularkan (Infectious).

````python
import numpy as np
from scipy.integrate import odeint
from scipy.optimize import minimize
import matplotlib.pyplot as plt

# 1. Definisikan Persamaan Diferensial SEIR
def SEIR_model(y, t, N, beta, sigma, gamma):
    S, E, I, R = y
    dSdt = -beta * S * I / N
    dEdt = beta * S * I / N - sigma * E
    dIdt = sigma * E - gamma * I
    dRdt = gamma * I
    return dSdt, dEdt, dIdt, dRdt

# 2. Fungsi Loss untuk Fitting Parameter SEIR (Mencari Beta, Sigma, Gamma)
def loss_function_seir(parameters, t, y_actual, N):
    beta, sigma, gamma = parameters

    # Kondisi awal hipotesis
    I0 = y_actual[0]
    R0 = actual_data['R_actual'].iloc[0]
    # Kita asumsikan kasus terpapar awal (E0) proporsional terhadap kasus aktif awal
    E0 = I0 * 1.5
    S0 = N - E0 - I0 - R0

    solution = odeint(SEIR_model, (S0, E0, I0, R0), t, args=(N, beta, sigma, gamma))
    I_pred = solution[:, 2]

    # Menggunakan MSE sebagai metrik evaluasi
    return np.mean((I_pred - y_actual) ** 2)

# 3. Optimasi Parameter SEIR
# Tebakan awal: beta=0.2, sigma=0.2 (inkubasi ~5 hari), gamma=0.1 (pemulihan ~10 hari)
initial_guess_seir = [0.2, 0.2, 0.1]
bounds_seir = ((0.001, 2.0), (0.01, 1.0), (0.001, 1.0))

result_seir = minimize(loss_function_seir, initial_guess_seir, args=(t_days, y_actual, N), bounds=bounds_seir, method='L-BFGS-B')
optimal_beta_seir, optimal_sigma_seir, optimal_gamma_seir = result_seir.x

print("Optimasi SEIR Selesai!")
print(f"Estimated Beta (laju penularan): {optimal_beta_seir:.4f}")
print(f"Estimated Sigma (laju inkubasi): {optimal_sigma_seir:.4f} (Masa inkubasi rata-rata: {1/optimal_sigma_seir:.1f} hari)")
print(f"Estimated Gamma (laju pemulihan): {optimal_gamma_seir:.4f}")
print(f"Basic Reproduction Number (R0): {optimal_beta_seir / optimal_gamma_seir:.2f}")
````

````python
# 4. Simulasi Hasil SEIR dengan Parameter Optimal
I0_seir = y_actual[0]
R0_seir = actual_data['R_actual'].iloc[0]
E0_seir = I0_seir * 1.5
S0_seir = N - E0_seir - I0_seir - R0_seir

sim_seir = odeint(SEIR_model, (S0_seir, E0_seir, I0_seir, R0_seir), t_days, args=(N, optimal_beta_seir, optimal_sigma_seir, optimal_gamma_seir))
S_seir, E_sim, I_seir, R_seir = sim_seir.T

# 5. Visualisasi Perbandingan SIR vs SEIR terhadap Data Aktual
plt.figure(figsize=(14, 8))
plt.plot(actual_data['Date'], y_actual, 'ro', label='Kasus Aktif Aktual (I)', alpha=0.4, markersize=4)
plt.plot(actual_data['Date'], I_sim, 'b--', label='Prediksi Model SIR', linewidth=2)
plt.plot(actual_data['Date'], I_seir, 'g-', label='Prediksi Model SEIR', linewidth=2.5)

plt.title('Perbandingan Performa Model SIR vs Model SEIR terhadap Data Aktual COVID-19', fontsize=16)
plt.xlabel('Tanggal', fontsize=12)
plt.ylabel('Jumlah Kasus Aktif', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(fontsize=12)
plt.show()

# Menghitung metrik evaluasi (RMSE)
mse_sir = np.mean((I_sim - y_actual)**2)
mse_seir = np.mean((I_seir - y_actual)**2)
print(f"Root Mean Squared Error (RMSE) SIR : {np.sqrt(mse_sir):,.2f}")
print(f"Root Mean Squared Error (RMSE) SEIR: {np.sqrt(mse_seir):,.2f}")
````

```markdown
## Rencana Analisis Lanjutan (Post-Modeling Epidemiologi)

Setelah mendapatkan estimasi parameter dasar untuk model SIR dan SEIR, berikut adalah beberapa langkah analisis strategis berikutnya yang dapat kita lakukan:

### 1. Simulasi Skenario Intervensi (Intervention & Policy Simulation)
Kita dapat memodifikasi parameter $\beta$ (laju penularan) pada titik waktu tertentu untuk menyimulasikan dampak kebijakan pemerintah, seperti:
* **Lockdown / Pembatasan Sosial (PSBB)**: Menurunkan $\beta$ sebesar 30%, 50%, atau 70% untuk melihat seberapa jauh puncak kurva kasus aktif dapat ditekan dan ditunda (*flattening the curve*).
* **Peningkatan Protokol Kesehatan / Penggunaan Masker**: Menyimulasikan penurunan transmisi harian secara bertahap.

### 2. Prediksi Proyeksi Masa Depan (Forecasting Beyond the Dataset)
Melakukan ekstrapolasi model SEIR hingga beberapa bulan ke depan melampaui tanggal akhir dataset (27 Juli 2020) untuk memperkirakan:
* Kapan puncak pandemi (*peak infection*) sesungguhnya akan tercapai jika tanpa intervensi.
* Kapan pandemi diperkirakan mereda (kasus aktif mendekati nol).

### 3. Analisis Parameter Spesifik Negara (Country-Specific Fitting)
Karena pola penyebaran di setiap negara berbeda akibat kebijakan dan fasilitas kesehatan:
* Kita bisa memilih beberapa negara dari 5 besar (misal: **US**, **Brazil**, atau **India**).
* Melakukan fitting model SEIR secara terpisah untuk membandingkan nilai $R_0$ (Reproduction Number) dan masa inkubasi antar-negara tersebut.

### 4. Sensitivitas Populasi Rentan (Sensitivity Analysis on Population $N$)
Meneliti bagaimana perubahan estimasi total populasi rentan ($N$) memengaruhi akurasi prediksi model dan pergeseran waktu puncak pandemi.
```
